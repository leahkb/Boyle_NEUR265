```python
#2. Make a code cell that imports numpy as np...

#Import all necessary libraries for the homework
import numpy as np
import matplotlib.pyplot as plt
import statistics as st
```

```python
#3. Make a code cell that imports your v_rest_sst.csv ...

#Extract data from urls using np.loadtxt

url_pvalb = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
url_sst = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'

v_rest1 = np.loadtxt(url_sst)
v_rest2 = np.loadtxt(url_pvalb)

print(type(v_rest1))
print(type(v_rest2))
```

```python
#4. Make a code cell that concatenates v_rest_sst ...

#Extract data into np array of 2x500 (I think it is supposed to be 2x13 but I'm not sure how that is supposed to be divided )

v_rest = np.array([v_rest1, v_rest2])

print(type(v_rest))
print(v_rest.shape)
```

```python
# 5. Make a code cell that prints the size of v_rest

#Print the size attribute

print(v_rest.size) #Size

print(v_rest.shape) #Dimensions
```

```python
#6. Make a code cell that returns the first row of v_rest

#Print the first row

print(v_rest[0])
```

```python
#7. Make a code cell that returns the sum of the values in the third column...

#Print the sum of the values in the 3rd column (index 2)

#Check
print(v_rest[0][2])
print(v_rest[1][2])
print(v_rest[0][2] + v_rest[1][2])

print('sum = ', v_rest[:,2].sum())

#For checking purposes
print (v_rest[0][0:5])
print (v_rest[1][0:5])
```

```python
#9. Create a code cell taht makes a histogram of the values...

#Create a histogram and label the axes

plt.hist(v_rest[0])
plt.xlabel("Membrane potential")
plt.ylabel("Count")
```

```python
#10. create a code cell that takes the mean of teh values in the first row...

#print the mean of each row

v_rest_mean_sst = st.mean(v_rest[0])
v_rest_mean_pvalb = st.mean(v_rest[1])

print(v_rest_mean_sst)
print(v_rest_mean_pvalb)
```

11.
The mean difference is only about 2 mV. This means that the pvalb average resting membrane potential is slightly more negative than the sst.


```python
#12. How can we test whether the difference in means ...

#import stats library

import scipy.stats as stats
```

```python
#13. Now, we can perform a t-test on the two rows ...

#perform a t-test and interpret the findings

stats = stats.ttest_ind(v_rest[0], v_rest[1])
print(stats.pvalue)
```

14. The results of the ttest say that the pvalue is less than 0.05 or 5%, which means that the results are statistically significant.
