```python
#2. Make a code cel that imports numpy as np...

#Import libraries for homework
import numpy as np
import matplotlib.pyplot as plt
import scipy.stats as stats

```

```python
#3. Make a code cel that imports...

#Import csv with data as numpy arrays

url_pvalb = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
url_sst = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'

v_rest_sst = np.loadtxt(url_sst)
v_rest_pvalb = np.loadtxt(url_pvalb)

print(type(v_rest_sst))
print(type(v_rest_pvalb))
```

```python
#4. Make a code cell that does the following:


#Finds the mean of v_rest_sst and v_rest_pvalb
mean_sst = np.mean(v_rest_sst)
mean_pvalb = np.mean(v_rest_pvalb)

print("sst mean:", mean_sst, " pvalb mean:", mean_pvalb)

#Check with statistics
import statistics as st

print(st.mean(v_rest_sst), st.mean(v_rest_pvalb))

#Preform a t-test to check if means come from different distributions
t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)
print("\n\n T-test pvalue result: ",  p_value, "< 0.05 ... so the means come from different distributions" if p_value <0.05 else "> 0.05")

#Plot a histogram of v_rest_sst and v_rest_pvalb
if p_value < 0.05:
  fig, ax = plt.subplots(1,1) #gives whole figure and axes variables
  ax.set_xlabel("Membrane potential")
  ax.set_ylabel("Count")

  ax.hist(v_rest_sst, alpha = 0.5, label = "sst") #alpha adjusts transparency
  ax.hist(v_rest_pvalb, alpha = 0.5, label = "pvalb")

  fig.set_facecolor("lightgrey") #experimenting with colors
  ax.legend()
  plt.show()
else:
  print("The p-value is not less than 0.05")
```

```python
#5. Make a code cell that defines a function ...

#Create vrestHist

def vrestHist():
  '''
  This function contains the code to display a histogram of sst and pvalb neurons
  if the p-value is less than 0.05 for the two datasets provided.
  '''
  #Preform a t-test to check if means come from different distributions
  t_stat, p_value = stats.ttest_ind(v_rest_sst, v_rest_pvalb)
  print("\n\n T-test pvalue result: ",  p_value, "< 0.05 ... so the means come from different distributions" if p_value <0.05 else "> 0.05")

  #Plot a histogram of v_rest_sst and v_rest_pvalb
  if p_value < 0.05:
    fig, ax = plt.subplots(1,1) #gives whole figure and axes variables
    ax.set_xlabel("Membrane potential")
    ax.set_ylabel("Count")

    ax.hist(v_rest_sst, alpha = 0.5, label = "sst") #alpha adjusts transparency
    ax.hist(v_rest_pvalb, alpha = 0.5, label = "pvalb")

    fig.set_facecolor("lightgrey") #experimenting with colors
    ax.legend()
    plt.show()
  else:
    print("The p-value is not less than 0.05")
```

```python
#6. Make a code cell that calls vrestHist()

vrestHist()
```
