```python
#2. Create a code cell that imports

import numpy as np
import matplotlib.pyplot as plt
```

```python
#3. upload data
url_trial1_left = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial1.csv"
url_trial1_right = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial1.csv"

url_trial2_left = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_left_trial2.csv"
url_trial2_right = "https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/spikes_right_trial2.csv"

spikes_left_trial1 = np.loadtxt(url_trial1_left, delimiter=',')
spikes_right_trial1 = np.loadtxt(url_trial1_right, delimiter=',')
spikes_left_trial2 = np.loadtxt(url_trial2_left, delimiter=',')
spikes_right_trial2 = np.loadtxt(url_trial2_right, delimiter=',')
```

```python
#4. print lenght of each array

print(len(spikes_left_trial1),
      len(spikes_right_trial1),
      len(spikes_left_trial2),
      len(spikes_right_trial2))
```

```python
#printing out the data

print(spikes_left_trial1[0:5])
print(spikes_right_trial1[0:5])
print(spikes_left_trial2[0:5])
print(spikes_right_trial2[0:5])
```

```python
#6. Write an expression to determine the firing rate of this neuron
#during each trial

t1_left = spikes_left_trial1[-1] #total time for each
t1_right = spikes_right_trial1[-1]
t2_left = spikes_left_trial2[-1]
t2_right = spikes_right_trial2[-1]

fr_t1l = len(spikes_left_trial1)/t1_left
fr_t1r = len(spikes_right_trial1)/t1_right
fr_t2l = len(spikes_left_trial2)/t2_left
fr_t2r = len(spikes_right_trial2)/t2_right

print(fr_t1l,fr_t1r, fr_t2l, fr_t2r)
```

7. There does not seem to be a pattern of which neurons fire faster (left or right) Based on what I found.

```python
#8. make subplots to display action potential data
fig, ax = plt.subplots(1,2, figsize=(10,5))

ax[0].plot(spikes_left_trial1, np.ones_like(spikes_left_trial1), '.', color = 'red')
ax[0].plot(spikes_right_trial1, 1.5 * np.ones_like(spikes_right_trial1), '.', color = 'blue')
ax[0].set_title("Trial 1")
ax[0].set_xlabel("Time (s)")
ax[0].set_ylabel("Turn direction")
ax[0].set_yticks([1, 2])
ax[0].set_yticklabels(["Left", "Right"])
ax[1].plot(spikes_left_trial2, np.ones_like(spikes_left_trial2), '.', color = 'red')
ax[1].plot(spikes_right_trial2, 1.5 * np.ones_like(spikes_right_trial2), '.', color = 'blue')
ax[1].set_title("Trial 2")
ax[1].set_xlabel("Time (s)")
ax[1].set_yticks([1, 2])
ax[1].set_yticklabels(["Left", "Right"])
```

9. This neuron seems to be more bursty during the left intervals than during the right intervals. These happen in red at the bottom of the graph, where there are larger groups of dots separated by more whitespace.

```python
#10. Calculate ISI for each array using np.diff
ISI_t1_l = np.diff(spikes_left_trial1)
ISI_t1_r = np.diff(spikes_right_trial1)
ISI_t2_l = np.diff(spikes_left_trial2)
ISI_t2_r = np.diff(spikes_right_trial2)

print("ISI Values: ", ISI_t1_l, ISI_t1_r, ISI_t2_l, ISI_t2_r)
```

```python
#11. make histograms to display ISI data

fig, ax = plt.subplots(2,2, figsize=(10,5))

bins = np.arange(0, 2, 0.01)

fig.suptitle("ISI for trials one and two", fontsize=14)
ax[0,0].hist(ISI_t1_l, bins)
ax[0,1].hist(ISI_t1_r, bins)
ax[1,0].hist(ISI_t2_l, bins)
ax[1,1].hist(ISI_t2_r, bins)

plt.subplots_adjust(hspace=0.5)

ax[0,0].set_xlabel('t1 left ISI [s]')
ax[1,0].set_xlabel('t1 right ISI [s]')
ax[0,1].set_xlabel('t2 left ISI [s]')
ax[1,1].set_xlabel('t2 right ISI [s]')

ax[0,0].set_ylabel('Counts')
ax[1,0].set_ylabel('Counts')
ax[0,1].set_ylabel('Counts')
ax[1,1].set_ylabel('Counts')

```

12. This neuron seems to fire with more burstiness when it goes to the left in trial one, but the right in trial 2.

```python
#13. Get the firing rate and plot them over time
fig,ax = plt.subplots(1,2, figsize=(15,5))

time_bins = np.arange(0, 30, 1)
fr_left_trial1, _ = np.histogram(spikes_left_trial1, time_bins)
fr_right_trial1, _ = np.histogram(spikes_right_trial1, time_bins)
fr_left_trial2, _ = np.histogram(spikes_left_trial2, time_bins)
fr_right_trial2, _ = np.histogram(spikes_right_trial2, time_bins)

ax[0].set_title("TRIAL 1")
ax[0].plot(time_bins[:-1], fr_left_trial1, '.')
ax[0].plot(time_bins[:-1], fr_right_trial1, '.')

ax[1].set_title("TRIAL 2")
ax[1].plot(time_bins[:-1], fr_left_trial2, '.')
ax[1].plot(time_bins[:-1], fr_right_trial2, '.')

ax[0].set_xlabel('Time [s]')
ax[0].set_ylabel('Number of Spikes')
ax[1].set_xlabel('Time [s]')
ax[1].set_ylabel('Number of Spikes')
```

14. The graph from step 13 displays the firing rate as count on the y axis and time on the x axis for trial one and two. The blue is the left and the orange is the right motion. Generally it shows that for trial one the left has more bursting and the right has more burst for trial 2. The values are pretty close for trial 2 though.
