```python
#2. Import the relevant modules

import numpy as np
import matplotlib.pyplot as plt
```

```python
#3. Import data

acc_fa = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_fa_hw5.csv", delimiter=",")
acc_hits = np.loadtxt("https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/acc_hits_hw5.csv", delimiter=",")

```

```python
#4. Print the shape of each array

print(acc_fa.shape)
print(acc_hits.shape)
```

```python
#6. Create the time variable with the same length as acc_hits and acc_fa
# with equally spaced values between 0 and 4

total_time = 4 #total time is 4 seconds
time = np.linspace(0, 4, acc_fa.shape[1])
print(time)
```

```python
#7. Determine the sampling rate of the data

num_samples = acc_fa.shape[1]
sr = num_samples/total_time
print(sr, "samples/second")
```

```python
#8. Plot the first trial of acc_hits in one subplot
#   Plot the first trial of acc_fa in another subplot

fig, ax = plt.subplots(1,2 , figsize=(15,5))
ax[0].plot(time, acc_hits[0])
ax[1].plot(time, acc_fa[1])

ax[0].set_title("Hit")
ax[0].set_xlabel("Time (s)")
ax[0].set_ylabel("Voltage (mV)")

ax[1].set_title("False Alarm")
ax[1].set_xlabel("Time (s)")
ax[1].set_ylabel("Voltage (mV)")
```

9. The dominant oscillation

```python
#10. Compare LFPs between fa and hits using ERP

plt.plot(time, acc_hits.mean(0), color = 'purple')
#plt.plot(time, acc_fa.mean(0), color = 'pink')
plt.xlabel('Time [s]')
plt.ylabel('Voltage [microV]')
plt.title('ERP of condition hits and fa')
plt.hlines(0, time[0], time[-1], color = 'black')
plt.vlines(2.0, -60 , 60, 'k', lw=1, color = 'black')
```

11.  The ERP results show that there is a change in activity after the time of touch ~2.0s. This suggests that thhere was activity after there was a hit.  

```python
#12. Repeat for acc_fa
#plt.plot(time, acc_hits.mean(0), color = 'purple')
plt.plot(time, acc_fa.mean(0), color = 'pink')
plt.xlabel('Time [s]')
plt.ylabel('Voltage [microV]')
plt.title('ERP of condition hits and fa')
plt.hlines(0, time[0], time[-1], color = 'black')
plt.vlines(2.0, -60 , 60, 'k', lw=1, color = 'black')
```

13. There does not appear to be a change in ERP behavior before or after the 2.0s time point where the mouse taps the screen. This suggests that the false alarm did not elicit a response in the LFP data.
