```python
#Reading csv with sst and pvalb data
url_sst = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_sst.csv'
pandas_data1 = pd.read_csv(url_sst)

url_pvalb = 'https://raw.githubusercontent.com/hallockh/neur_265_spring2026/refs/heads/main/data/v_rest_pvalb.csv'
pandas_data2 = pd.read_csv(url_pvalb)
```

```python
#Convert csv data to list types
data1 = pandas_data1.values.tolist()
data2 = pandas_data2.values.tolist()
```

```python
#Flatten data1 and data2 (remove brackets surrounding values)
data1 = np.squeeze(data1)
data2 = np.squeeze(data2)
```

```python
#Print the length of data1
print(len(data1))
```

```python
#Print the 3rd, 4th, 5th, 6th element of data2
#print(data2[2], data2[3], data2[4], data2[5])
print(data2[2:6])
```

```python
#Print the second-to-last and last elements of data1
print(data1[-2], data1[-1])
```

```python
#Print the type of data2
print(type(data2))
```

```python
#Create a histogram of data1
plt.hist(data1)
```

```python
#Remake the plot with proper labels
plt.ylabel('Count')
plt.xlabel('Membrane Potential (mV)')
plt.hist(data1)
```

```python
#Create a plot for data2
plt.ylabel('Count')
plt.xlabel('Membrane Potential (mV)')
plt.hist(data2)
```

The data1 set for SST cells has a resting membrane potential most frequently at -70 mV for all of the cells measured.
